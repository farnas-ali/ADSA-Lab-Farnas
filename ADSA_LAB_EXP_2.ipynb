```python
# Name: Mohammed Farnas Ali Mudabbir
# Mtech-CSE #RollNo: 2503B05136

import time  # to simulate time delay for check-up rounds

class PatientNode:
    def __init__(self, name, age, patient_id):
        self.name = name
        self.age = age
        self.patient_id = patient_id
        self.next = None

class CircularPatientList:
    def __init__(self):
        self.head = None
        self.size = 0  # <--- NEW: Track the number of patients

    def insert_patient(self, name, age, patient_id):
        new_node = PatientNode(name, age, patient_id)
        if self.head is None:
            self.head = new_node
            new_node.next = self.head
        else:
            current = self.head
            # Traverse until the node whose 'next' is the head
            while current.next != self.head:
                current = current.next
            current.next = new_node
            new_node.next = self.head
        
        self.size += 1 # <--- NEW: Increment size
        print(f"Inserted: {name}, Age: {age}, ID: {patient_id} (Size: {self.size})")

    def delete_patient(self, patient_id):
        if self.head is None:
            print("No patients to delete.")
            return

        current = self.head
        previous = None
        
        # Traverse exactly 'self.size' times to prevent infinite loop if structure breaks
        for _ in range(self.size): 
            if current.patient_id == patient_id:
                # --- START IMPROVED DELETION LOGIC ---
                if previous is not None:
                    # Case 1: Deleting an intermediate or tail node
                    previous.next = current.next
                else:
                    # Case 2: Deleting the head node
                    
                    # Subcase 2a: Single node list
                    if self.head == self.head.next:
                        self.head = None
                        self.size = 0 # Decrement size here as list is now empty
                        print(f"Deleted last patient with ID: {patient_id}")
                        return

                    # Subcase 2b: Head of a multi-node list
                    temp_tail = self.head
                    while temp_tail.next != self.head:
                        temp_tail = temp_tail.next
                        
                    temp_tail.next = self.head.next
                    self.head = self.head.next
                
                # --- END IMPROVED DELETION LOGIC ---
                self.size -= 1 # Decrement size for multi-node cases
                print(f"Deleted patient with ID: {patient_id} (Size: {self.size})")
                return
                
            previous = current
            current = current.next
            
            # Since we use self.size loop, this check is technically redundant 
            # but kept for original structure robustness.
            if current == self.head: 
                break

        print(f"Patient with ID {patient_id} not found.")

    def display_patients(self):
        if self.head is None:
            print("No patients in the list.")
            return

        print(f"\nPatient List (Round-Robin Order - Total: {self.size}):")
        current = self.head
        for _ in range(self.size): # <--- Using size for safe traversal
            print(f"Name: {current.name}, Age: {current.age}, ID: {current.patient_id}")
            current = current.next

    def simulate_checkups(self, rounds=2, delay=1.5):
        """Simulate round-robin patient check-ups."""
        if self.head is None:
            print("No patients to check.")
            return

        print("\n Starting Round-Robin Check-Up Simulation...\n")
        current = self.head
        for round_num in range(1, rounds + 1):
            print(f"--- Round {round_num} ---")
            
            for _ in range(self.size): # <--- Using size for safe traversal
                print(f"Now checking: {current.name} (Age: {current.age}, ID: {current.patient_id})")
                time.sleep(delay)  # simulate time gap between checkups
                current = current.next
                
            print()
        print("✅ Check-up simulation completed.\n")


# Example usage
if __name__ == "__main__":
    plist = CircularPatientList()

    # Insert 5 patients (custom order)
    plist.insert_patient("Alice", 30, 101)
    plist.insert_patient("Charlie", 25, 102)
    plist.insert_patient("Daniel", 50, 103)
    plist.insert_patient("Bob", 45, 104)
    plist.insert_patient("Fiona", 29, 105)

    # Display patients
    plist.display_patients()

    # Delete two patients
    plist.delete_patient(102)  # Charlie (intermediate)
    plist.delete_patient(101)  # Alice (head)

    # Display updated list
    plist.display_patients()

    # Simulate Round-Robin Check-Up for remaining patients
    plist.simulate_checkups(rounds=3, delay=1)
    
    # Test remaining deletions
    plist.delete_patient(105) # Fiona (tail)
    plist.delete_patient(103) # Daniel (intermediate)
    plist.delete_patient(104) # Bob (last remaining node)
    
    plist.display_patients()
```
